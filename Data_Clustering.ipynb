```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
# 1. Load the original dataset
df = pd.read_csv("london_airbnb.csv")
```

```python
# 2. Initial Cleaning
# Fill missing reviews with 0
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
```

```python
# Ensure all listings have spatial coordinates for clustering
df = df.dropna(subset=["latitude", "longitude"])
```

```python
# 3. Spatial Clustering
# Group listings into 5 clusters based on latitude and longitude
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(df[["latitude", "longitude"]])
```

```python
# 4. Feature Selection
# Drop columns that are entirely or mostly null (neighbourhood_group and license)
df = df.drop(columns=["neighbourhood_group", "license"])
```

```python
# 5. Price Outlier Capping
# Calculate the 99th percentile and keep listings between £10 and that cap
price_cap = df["price"].quantile(0.99)
df = df[df["price"].between(10, price_cap)]
```

```python
# 6. Final Formatting for Tableau/Visualization
# Create a friendly label for the clusters
df["cluster_label"] = "Cluster " + df["cluster"].astype(str)
```

```python
# 7. Save the final cleaned and clustered dataset
df.to_csv("london_airbnb_final.csv", index=False)
```
